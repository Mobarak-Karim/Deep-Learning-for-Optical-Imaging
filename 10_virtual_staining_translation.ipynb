# 10. Virtual Staining and Image-to-Image Translation

Virtual staining is not merely a colorization task. The scientific question is whether morphology and diagnostic/biological information are preserved without unsupported structures.

## Paired vs unpaired

Paired training requires meaningful spatial correspondence between input and target. Misregistration can teach blur or false transformations. Unpaired approaches relax pairing but make content preservation harder to guarantee and validate.

```python
import torch
from torch import nn

# Example: one-channel optical input -> three-channel virtual color output.
translator = nn.Sequential(
    nn.Conv2d(1, 32, 3, padding=1), nn.ReLU(),
    nn.Conv2d(32, 32, 3, padding=1), nn.ReLU(),
    nn.Conv2d(32, 3, 1), nn.Sigmoid(),
)
x = torch.rand(2,1,64,64)
print(translator(x).shape)
```

## Validation hierarchy

1. registration/pair quality;
2. color/intensity similarity;
3. structural similarity;
4. feature-level morphology;
5. blinded expert/pathology evaluation when relevant;
6. task performance and domain-shift testing.

A visually convincing virtual stain can still be scientifically wrong.
