# 11. Small Biomedical Datasets and Transfer Learning

Limited data changes the experimental design. It is not automatically solved by augmentation or a larger pretrained model.

## First questions

- How many independent specimens, not images/patches?
- Is there class/specimen imbalance?
- Does augmentation preserve the optical/biological meaning?
- Is ImageNet pretraining relevant to the input representation?
- Would cross-validation better characterize uncertainty?

## Freeze or fine-tune?

A common experiment is: train only the new head first, then compare with partial/full fine-tuning. Do not assume one strategy is universally best.

For grayscale optical images, explicitly decide whether to replicate channels, modify the first convolution, or train a modality-specific encoder.
