
## 1. "Many patches" does not mean "large biomedical dataset"

A dataset with 100,000 patches from 20 patients is still fundamentally limited by the diversity of 20 patients.

Report both:
- number of images/patches;
- number of independent biological units.


## 2. Transfer learning

Transfer learning starts from weights learned on another task/dataset.

Potential benefits:
- faster convergence;
- useful generic low-level features;
- less data required.

Potential mismatch:
- RGB natural-image pretraining vs single-channel OCT/fluorescence;
- domain statistics differ;
- resolution/texture differs.

Treat transfer learning as a hypothesis to validate, not a guaranteed improvement.


## 3. Freeze → unfreeze strategy

One common workflow:

```text
1. replace task-specific head
2. freeze encoder
3. train new head briefly
4. unfreeze part/all encoder
5. fine-tune with lower learning rate
```

But segmentation encoders may need more adaptation than simple classification heads.


## 4. Cross-validation

When independent sample count is small, k-fold cross-validation can estimate variability.

Important:
- split by specimen/patient;
- keep preprocessing fit within each fold;
- do not tune hyperparameters on the final held-out external test set.

Nested CV may be needed for rigorous hyperparameter selection, but it can be computationally expensive.


## 5. Data augmentation as invariance

Augmentation should encode transformations under which the label should remain valid.

Ask:
- Is rotation scientifically valid?
- Is flip valid?
- Is intensity scaling valid?
- Is blur/noise within acquisition range?
- Does deformation preserve pathology?

Augmentation is not a substitute for missing biological diversity.


## 6. Regularization

Common tools:
- weight decay;
- dropout;
- data augmentation;
- early stopping;
- smaller model;
- reduced input resolution/complexity;
- transfer learning.

The best regularizer is sometimes a better dataset split and a simpler model.


## 7. Class imbalance

For rare disease/features:
- report class counts by independent subject;
- use balanced sampling carefully;
- consider class-weighted loss;
- use AUROC **and** precision-recall when prevalence is low;
- report sensitivity/specificity at meaningful thresholds.

Do not artificially balance the test set without explaining how metrics relate to real prevalence.


## 8. Small-data decision tree

```text
Few independent subjects?
    ↓
simple baseline first
    ↓
specimen-level CV
    ↓
transfer learning baseline
    ↓
reasonable augmentation
    ↓
regularization
    ↓
uncertainty/confidence intervals
    ↓
external validation if possible
```
