## Reproduction ladder

```text
environment installs
→ imports succeed
→ one sample loads
→ one forward pass works
→ pretrained inference works
→ metric script works
→ one reported qualitative/quantitative behavior is approximated
→ only then adapt
```

## Shape/range checkpoint

Write these checks into temporary debugging code rather than relying on assumptions.

````python
import torch

def inspect_batch(x, y=None):
    print('x shape:', tuple(x.shape), 'dtype:', x.dtype,
          'range:', (float(x.min()), float(x.max())))
    if y is not None:
        print('y shape:', tuple(y.shape), 'dtype:', y.dtype,
              'range:', (float(y.min()), float(y.max())))

x = torch.rand(2, 1, 64, 64)
y = torch.rand(2, 1, 64, 64)
inspect_batch(x, y)
````
