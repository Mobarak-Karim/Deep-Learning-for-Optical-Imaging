
## 1. Reproduction comes before adaptation

Your first goal is **not** to improve the authors' code. It is to answer:

> Can I reproduce one result or behavior under conditions close to the original?

Only after that should you change architecture, data, losses, or versions.


## 2. Safe reproduction workflow

```text
1. Record paper + repository URL + commit/tag
2. Read license
3. Create isolated environment
4. Install closest documented dependency versions
5. Obtain the intended data/weights legally
6. Run authors' smallest inference example
7. Run one evaluation example
8. Reproduce one reported metric/figure approximately
9. Save logs and deviations
10. Only then create an adaptation branch/environment
```


## 3. Freeze the reference point

Record a commit hash/tag. A repository can change after publication.

Also save:
- config file;
- command used;
- environment package list;
- model checkpoint checksum/name;
- dataset version;
- any manual fixes.


## 4. Make the smallest possible fix

If old code fails:
1. capture the exact error;
2. identify whether it is environment/API/data related;
3. make one minimal change;
4. document it;
5. retry.

Do not simultaneously upgrade PyTorch, rewrite the dataloader, change normalization, and switch GPUs. You will lose the ability to attribute differences.


## 5. Reproduction levels

A useful hierarchy:

| Level | Evidence |
|---|---|
| 0 | repository imports |
| 1 | provided inference example runs |
| 2 | released checkpoint reproduces example output |
| 3 | evaluation metric approximately matches paper |
| 4 | training from provided recipe reaches similar behavior |
| 5 | method generalizes to a second dataset/domain |

You may not need Level 4 for every project. State which level you achieved.


## 6. Reproduction log template

```text
Paper:
Repo:
Commit:
Date:
Environment:
GPU:
Dataset:
Checkpoint:
Command:
Expected:
Observed:
Difference:
Changes made:
Result:
Next action:
```


## 7. When a metric does not match

Check in this order:
- exact test split;
- preprocessing;
- checkpoint;
- inference resize/tiling;
- threshold/postprocessing;
- metric implementation;
- aggregation level;
- package/model behavior changes;
- random seed/training variation.

Do not assume the architecture is wrong first.


## 8. Optical-imaging example

Suppose a MUSE-to-H&E paper provides inference code and weights but not training data.

A reasonable reproduction can still be:
1. verify license;
2. run released weights on provided example images;
3. reproduce example color/morphology behavior;
4. inspect preprocessing;
5. test on a small legally available/authorized sample from your domain;
6. treat cross-domain performance as **adaptation/generalization**, not reproduction.
