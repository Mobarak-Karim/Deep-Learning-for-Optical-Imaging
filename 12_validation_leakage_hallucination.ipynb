
## 1. Validation asks a scientific question

The goal is not "Does my model get a high metric?"

The goal is:

> Under what conditions does the model make reliable predictions, and where does it fail?


## 2. Leakage taxonomy

### Direct leakage
Same image/sample in train and test.

### Group leakage
Different patches/slices from same patient/animal in different splits.

### Preprocessing leakage
Normalization/PCA/statistics fit using all data including test.

### Label leakage
Filename, annotation marks, scale bars, metadata, acquisition artifact correlates with label.

### Temporal/site leakage
Train/test separation accidentally follows acquisition settings rather than biology.


## 3. Internal versus external validation

- **Internal test:** same dataset/source distribution, unseen specimens.
- **External test:** different site, instrument, cohort, acquisition period, or protocol.

External validation measures a different and often more important question: domain generalization.


## 4. Confidence intervals and variability

A single average hides uncertainty.

For subject-level metrics, bootstrap subjects (not individual correlated pixels/patches) when appropriate to estimate uncertainty.

Report distribution/median/IQR or confidence intervals where useful.


## 5. Calibration

For probabilistic classification, calibration asks whether predictions such as 0.8 confidence correspond to about 80% empirical correctness.

A model can have good AUROC but poor calibration.


## 6. Uncertainty is not one thing

Sources include:
- aleatoric: intrinsic noise/ambiguity;
- epistemic: model uncertainty due to limited knowledge/data;
- distribution shift: sample unlike training domain.

Methods such as ensembles/MC dropout can estimate aspects of uncertainty, but no uncertainty score guarantees safety.


## 7. Hallucination testing for image generation/restoration

Use controlled experiments:
- remove a known structure from input;
- insert synthetic structure with known amplitude;
- vary SNR systematically;
- test blank/background fields;
- test phantoms;
- compare downstream measurements before/after.

Ask whether model error is **biased toward plausible anatomy**.


## 8. Failure-case library

Save examples for:
- best cases;
- median cases;
- worst cases;
- low SNR;
- rare pathology/morphology;
- artifacts;
- out-of-distribution acquisitions.

A paper-quality model report should make failures visible, not just show attractive examples.


## 9. Optical-imaging domain shifts to test

- different microscope/scanner;
- objective/NA;
- detector;
- wavelength;
- pixel size;
- exposure/power;
- reconstruction software/settings;
- staining batch;
- tissue preparation;
- acquisition day;
- operator/site;
- disease severity/population.


## 10. Validation checklist

```text
□ Split unit = independent biological unit
□ No preprocessing fit on test
□ Validation used for model selection only
□ Test used after decisions frozen
□ Per-specimen metrics saved
□ Confidence/variability reported
□ Worst cases inspected
□ Domain-shift tests designed
□ Hallucination tests for restoration/generation
□ External validation where feasible
□ Scientifically meaningful downstream measures checked
```
