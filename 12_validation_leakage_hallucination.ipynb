# 12. Validation, Leakage, Hallucination, and Uncertainty

This is the scientific core of the course. A high metric can coexist with an invalid experiment.

## Leakage checklist

- patches from the same specimen in multiple splits;
- normalization statistics calculated using test data;
- repeated test-set inspection during model design;
- augmentation before specimen-level split;
- near-duplicate frames across train/test;
- pretrained model trained on hidden overlapping data.

## Hallucination question

For restoration/translation ask two separate questions:

1. **Deletion:** can the model remove a real structure?
2. **Insertion:** can it add a structure not supported by the measurement?

Design challenge sets that contain faint, rare, small, or out-of-distribution structures.
