# 14. Final Project — Optical-Imaging Paper → Audited Code → Validated Baseline

Choose one deep-learning paper related to optical imaging. The deliverable is not merely a trained model; it is a defensible implementation record.

## Deliverable 1 — Paper specification

Complete `PAPER_TO_CODE_CHECKLIST.md`. Extract input, target, split, preprocessing, architecture, equations/losses, optimizer, metrics, and reported limitations.

## Deliverable 2 — Code decision

Audit any shared repository. Choose and justify **reuse / adapt / reimplement / reject**. Include license and exact commit/tag.

## Deliverable 3 — Reproduction gate

Before modification, reproduce the smallest meaningful original result: one pretrained inference, one reported example, or one metric behavior.

## Deliverable 4 — Your baseline

Implement the smallest scientifically meaningful baseline for your own data. Include tensor/range checks and a tiny-dataset overfit test.

## Deliverable 5 — Validation

Use specimen-level splitting. Define at least one task metric and one optical/biological validity check. For restoration/translation explicitly test deletion/insertion of structures.

## Deliverable 6 — Optimization

Measure baseline throughput/memory. Apply only justified optimizations (DataLoader, AMP, profiler-guided changes, `torch.compile` when beneficial). Report before/after measurements.

## Final report template

```text
Scientific question:
Paper and DOI:
Authors' repository + commit:
License:
Decision (reuse/adapt/reimplement/reject):
Reason:
Data and specimen-level split:
Baseline:
Reproduction result:
Validation metrics:
Optical/biological validation:
Failure cases:
Optimization before/after:
Known limitations:
Next experiment:
```

A successful final project is one another researcher can understand, rerun, criticize, and improve.
