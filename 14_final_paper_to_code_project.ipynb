
## 1. Final project goal

Choose one optical-imaging paper/method and produce a complete evidence trail from scientific question to implementation decision.

Your deliverable is not just a trained model. It is a **reproducible research package**.


## 2. Stage A — Define your problem

Write:

```text
Scientific question:
Modality:
Independent biological unit:
Input:
Target:
Primary metric:
Scientific failure that matters most:
Deployment/research domain:
```

Example:
```text
Question: Can a model segment embryonic vessels from fluorescence images?
Unit: embryo
Input: single-channel 2-D image
Target: vessel mask
Metric: vessel Dice + sensitivity
Critical failure: deletion of small vessels
```


## 3. Stage B — Select and read the paper

Complete:
- `PAPER_TO_CODE_CHECKLIST.md`
- architecture table;
- loss equations;
- preprocessing table;
- split description;
- metric definition;
- discrepancy log.


## 4. Stage C — Audit the authors' code

Choose:
- REUSE
- ADAPT
- REIMPLEMENT
- REJECT

Write the evidence for the decision using `CODE_REUSE_DECISION_GUIDE.md`.


## 5. Stage D — Reproduction baseline

Before your own data:
- freeze commit/environment;
- run example;
- reproduce one reported behavior/metric if possible;
- record deviations.

If reproduction is impossible, explain exactly why.


## 6. Stage E — Build your own simple baseline

Your baseline should be simpler than the paper when possible.

Examples:
- U-Net before transformer segmentation;
- residual CNN before diffusion restoration;
- deterministic paired translator before GAN;
- ResNet-18 before a large ViT classifier.

This tells you whether paper complexity adds value.


## 7. Stage F — Validate scientifically

At minimum:
- specimen-level split;
- tiny-set overfit test;
- learning curves;
- primary + secondary metrics;
- per-specimen results;
- visual QC;
- failure cases;
- domain-shift test if feasible;
- hallucination/deletion stress tests for generated/restored images.


## 8. Stage G — Optimize after correctness

Record baseline time/memory, then test one optimization at a time:
- AMP;
- loader settings;
- batch strategy;
- compile;
- smaller input/model if scientifically acceptable.

Keep the validation metric in the optimization table.


## 9. Final report structure

```text
1. Scientific question
2. Paper selected and why
3. Paper specification
4. Repository/license audit
5. Reproduction outcome
6. Reuse/adapt/reimplement decision
7. Baseline implementation
8. Data split and preprocessing
9. Training details
10. Validation results
11. Failure/hallucination analysis
12. Optimization results
13. Limitations
14. What would be required before real scientific/clinical use?
```


## 10. Graduation test

You have completed the course when you can look at an unfamiliar deep-learning optical-imaging paper and answer, with evidence:

1. What problem does it actually solve?
2. Is the split scientifically valid?
3. What exact data transformation reaches the model?
4. What exact objective is optimized?
5. Does the public code match the paper?
6. Can you reproduce a reference result?
7. Should you reuse, adapt, reimplement, or reject?
8. What baseline should you build first?
9. What failures could invalidate the result?
10. How would you optimize only after correctness?
